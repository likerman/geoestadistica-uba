# Semejanza y agrupamiento multivariado

Ejemplo sintético para explorar cómo escala, distancia y algoritmo definen los grupos obtenidos.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.cluster.vq import kmeans2

datos = pd.DataFrame({'muestra':list('ABCDEFGHI'), 'porosidad':[8.2,8.8,9.1,14.5,15.2,14.8,22,22.8,21.5], 'Sr_ppm':[410,430,395,600,630,580,760,790,735]})
X = datos[['porosidad','Sr_ppm']].to_numpy()
Z = (X-X.mean(0))/X.std(0,ddof=1)
D = pd.DataFrame(squareform(pdist(Z)), index=datos.muestra, columns=datos.muestra)
D.round(2)
```

```python
L = linkage(Z, method='ward')
grupos_h = fcluster(L, 3, criterion='maxclust')
centros, grupos_k0 = kmeans2(Z, 3, minit='++', seed=2026)
grupos_k = grupos_k0 + 1
pd.DataFrame({'muestra':datos.muestra, 'jerarquico':grupos_h, 'kmeans':grupos_k})
```

```python
fig, axes = plt.subplots(1,2,figsize=(10.5,4.2),constrained_layout=True)
dendrogram(L, labels=datos.muestra.tolist(), ax=axes[0], color_threshold=None)
axes[0].set(title='Ward sobre variables estandarizadas',ylabel='Aumento de inercia')
axes[1].scatter(Z[:,0],Z[:,1],c=grupos_k,cmap='Set2',s=75,edgecolor='black')
for i,m in enumerate(datos.muestra): axes[1].text(Z[i,0]+.04,Z[i,1]+.04,m)
axes[1].set(xlabel='Porosidad (z)',ylabel='Sr (z)',title='k-means, K=3')
plt.show()
```

## Para interpretar

Repita sin estandarizar y con otros enlaces. Un grupo es una construcción dependiente de variables, escala y algoritmo; no constituye por sí mismo una facies o población geológica verdadera.
